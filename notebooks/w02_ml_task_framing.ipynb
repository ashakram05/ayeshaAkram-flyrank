# Week 2 – ML Task Framing

## My Lane
**Lane:** Decline Recovery

## 1. My Lane as an ML Task

**Task Type:** Ranking

The goal is to rank content pages by review priority so the SEO/content team knows which pages should be reviewed first for possible performance decline.

## 2. Target (or Proxy)

Ideally, I would like the model to predict whether a page truly needs a content refresh.

However, datasets rarely contain a direct label such as "Needs Review."

Instead, I use a proxy target.

In the starter dataset, the column `trend_direction` indicates whether a page's recent search trend is moving up, down, or remaining stable. A page with a value of `"down"` can reasonably represent a page that may require attention.

Therefore, I create a binary target called `is_declining_label`, where:

- 1 = trend is down
- 0 = otherwise

Although this is not a perfect representation of "needs review," it provides a practical target for training and evaluation.

## 3. Success Metric

The success metric for this task is **Precision@K**.

The content team has limited time and cannot review every page. Instead, they only investigate the highest-priority recommendations generated by the model.

Precision@K measures how many of the top K recommended pages are actually experiencing decline.

For example, if the model recommends the top 20 pages for review and 18 truly need attention, the Precision@20 is:

18 / 20 = 0.90

This metric directly reflects how useful the ranking is for the people using it.

```
import pandas as pd

# Load the anonymized FlyRank dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create a binary target from trend direction
df["is_declining_label"] = (
    df["trend_direction"].str.lower() == "down"
).astype(int)

# Display the first few rows
df.head()
```

## 4. Unit of Analysis

**One row = one content page.**

Each row contains signals such as impressions, CTR, average position, content age, and days since last update.

```
print(df.shape)
df[["impressions_90d","ctr","avg_position","days_since_last_update","content_age_days","is_declining_label"]].head()
```

## 5. Why ML Beats a Fixed Rule

A simple rule could be:

> Review every page that has not been updated for more than 180 days.

While this rule is easy to understand, it ignores many other important signals.

For example, an old page may still have strong search performance, while a newer page may already be losing visibility.

Machine learning considers multiple features at the same time, including impressions, CTR, search ranking, content age, and update history. By learning relationships between these signals, it can produce a more reliable ranking than a single manually written rule.

This makes ML more flexible and better suited for complex real-world data.

## 6. Real Business Action

The ranked output helps the content team:
- Prioritize which pages to review first.
- Spend editing time on pages with the highest risk of decline.
- Improve search performance more efficiently.
